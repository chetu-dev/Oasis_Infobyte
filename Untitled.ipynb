```python
# ## Data Quality Report - Initial Observations

# The initial data quality assessment was performed to identify missing values,
# duplicate records, data types, and potential value inconsistencies.

# The dataset contains 12 columns representing passenger information,
# survival status, demographic information, ticket details, and fare.

# Missing values, if present, will be handled using an appropriate strategy
# for each column. Duplicate records will be investigated before removal.
# Categorical values and numerical ranges will also be checked for
# inconsistencies and anomalies.
```

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Age'].isnull().sum()
```

```python
df['Cabin'] = df['Cabin'].fillna('Unknown')
```

```python
df['Cabin'].isnull().sum()
```

```python
df['Cabin'].isnull().sum()
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df['Embarked'].isnull().sum()
```

```python
df['Embarked'].isnull().sum()
```

```python
# ## Missing Data Handling

# Missing values were handled using column-specific strategies.

# - Age: Missing values were replaced with the median age because Age is a numerical variable and the median is less affected by extreme values.
# - Cabin: Missing values were replaced with "Unknown" because Cabin is a categorical variable and the missing information should not be artificially created.
# - Embarked: Missing values were replaced with the mode because Embarked is categorical and only a small number of values were missing.
# - Columns without missing values were left unchanged.

# After applying these strategies, the dataset was checked again to confirm that missing values were handled.
```

```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
age_outliers = df[
    (df['Age'] < lower_limit) |
    (df['Age'] > upper_limit)
]

print("Number of Age outliers:", len(age_outliers))
```
