```python
#[ ] Feature selection: identify 2–3 key behavioural features for clustering (e.g., Recency, Frequency, Monetary — RFM analysis)
```

```python
df[['Customer_ID', 'Date', 'Total_Amount']].head()
```

```python
df['Customer_ID'].nunique()
```

```python
df['Date'].min(), df['Date'].max()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
reference_date = df['Date'].max() + pd.Timedelta(days=1)

reference_date
```

```python
rfm = df.groupby('Customer_ID').agg(
    Recency=('Date', lambda x: (reference_date - x.max()).days),
    Frequency=('Order_ID', 'count'),
    Monetary=('Total_Amount', 'sum')
)

rfm.head()
```

```python
rfm.shape
```

```python
rfm.describe()
```

```python
rfm.isnull().sum()
```

```python
rfm
```

```python
#[ ] Data normalisation/standardisation before clustering (use StandardScaler)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
rfm_data = rfm[['Recency', 'Frequency', 'Monetary']]
```

```python
rfm_data.head()
```

```python
scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(rfm_data)
```

```python
rfm_scaled[:5]
```

```python
#[ ] Apply K-Means clustering — use the Elbow Method to determine optimal number of clusters (K)
    
```

```python
from sklearn.cluster import KMeans
```

```python
wcss = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(rfm_scaled)
    wcss.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(range(2, 11), wcss, marker='o')

plt.title('Elbow Method for Optimal Number of Clusters')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS')

plt.show()
```

```python
n_clusters = 4
```

```python
k = 4

kmeans = KMeans(
    n_clusters=k,
    random_state=42,
    n_init=10
)
```

```python
rfm['Cluster'] = kmeans.fit_predict(rfm_scaled)
```

```python
rfm.head()
```

```python
rfm['Cluster'].value_counts().sort_index()
```

```python
cluster_profile = rfm.groupby('Cluster')[['Recency', 'Frequency', 'Monetary']].mean()

cluster_profile
```

```python
#[ ] Visualise clusters using scatter plots (at least 2 feature combinations)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    rfm['Recency'],
    rfm['Frequency'],
    c=rfm['Cluster'],
    cmap='viridis',
    alpha=0.6
)

plt.title('Customer Segments: Recency vs Frequency')
plt.xlabel('Recency')
plt.ylabel('Frequency')
plt.colorbar(label='Cluster')

plt.show()
```

```python
#[ ] Profile each cluster: calculate mean feature values per cluster and describe the customer type
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    rfm['Frequency'],
    rfm['Monetary'],
    c=rfm['Cluster'],
    cmap='viridis',
    alpha=0.6
)

plt.title('Customer Segments: Frequency vs Monetary')
plt.xlabel('Frequency')
plt.ylabel('Monetary')
plt.colorbar(label='Cluster')

plt.show()
```
